```python
#################
#University of Toronto
#Faculty of Information
#Master of Information Program
#INF 1340H - Program for Data Science
#
#Student Name: Ruiqi
#Student Number: 1006743419
#Supervisor: Dr. Maher Elshakankiri
#
#
#Assignment 2, Problem 1
#Purpose: Purchases
#Date Created: 10.1.2026
#Date Modified: 10.1.2026
#################


def item_quantity_input() -> int:  # Request for item quantity input
    while True:
        quantity = input("Enter quantity: ")
        try:
            quantity = int(quantity)
            if quantity > 0:
                break
            else:
                print("Please input a positive integer.")
        except ValueError:
            print("Please input a positive integer.")
    return quantity


def unit_cost_input() -> float:  # Request for unit cost input
    while True:
        unit_cost = input("Enter unit cost: ")
        try:
            unit_cost = float(unit_cost)
            if unit_cost > 0:
                break
            else:
                print("Please input a positive number.")
        except ValueError:
            print("Please input a positive number.")
    return unit_cost


def is_taxable_input() -> bool:  # Request for is item taxable input
    while True:
        taxable = input("Is this item taxable (y/n)? ")
        if taxable in ("y", "n"):
            break
        else:
            print("Please input y or n.")
    if taxable == "y":
        return True
    else:
        return False


def each_item_input() -> list[int, float, bool]:  # Create a list for each item
    item_information = [item_quantity_input(), unit_cost_input(), is_taxable_input()]
    return item_information


def tax_rate_input() -> float:  # Request for tax rate
    while True:
        tax_rate = input("Enter sales tax rate (%): ")
        try:
            tax_rate = float(tax_rate)
            if tax_rate > 0:
                break
            else:
                print("Please input a positive number.")
        except ValueError:
            print("Please input a positive number.")
    return tax_rate


def number_of_items_input() -> int:  # Request for number of items
    while True:
        number_of_item = input("Enter number of different items: ")
        try:
            number_of_item = int(number_of_item)
            if number_of_item > 0:
                break
            else:
                print("Please input a positive integer.")
        except ValueError:
            print("Please input a positive integer.")
    return number_of_item


def main():  # Main program
    purchase_detail = []
    for i in range(1, number_of_items_input() + 1):
        item_name = "Item " + str(i)
        print(item_name)
        item_detail = [item_name] + each_item_input()
        purchase_detail.append(item_detail)
    tax_rate = tax_rate_input()
    total_cost = 0
    taxable_cost = 0
    for item_detail in purchase_detail:
        cost = item_detail[1] * item_detail[2]
        total_cost += cost
        if item_detail[3]:
            taxable_cost += cost
    tax = taxable_cost * 0.01 * tax_rate
    total_cost_after_tax = total_cost + tax
    print(f"Total cost (pre-tax): $ {total_cost:.2f}")
    print(f"Sales tax: $ {tax:.2f}")
    print(f"Total cost (with tax): $ {total_cost_after_tax:.2f}")


if __name__ == "__main__":
    main()
```

```python
#################
#University of Toronto
#Faculty of Information
#Master of Information Program
#INF 1340H - Program for Data Science
#
#Student Name: Ruiqi
#Student Number: 1006743419
#Supervisor: Dr. Maher Elshakankiri
#
#
#Assignment 2, Problem 2
#Purpose: Mean and Standard Deviation Excluding Outliers
#Date Created: 10.1.2026
#Date Modified: 10.1.2026
#################


def number_list_input() -> list[float]:  # Request for number list
    while True:
        number_list = input("Enter the list of numbers separated by spaces: ")
        try:
            number_list = number_list.strip().split()
            for i in range(len(number_list)):
                number_list[i] = float(number_list[i])
            break
        except ValueError:
            print("Please enter the list of numbers separated by spaces")
    return number_list


def range_input() -> list[float]:  # Request for range
    while True:
        low_range = input("Enter the low value of the range: ")
        high_range = input("Enter the high value of the range: ")
        try:
            low_range, high_range = float(low_range), float(high_range)
            if low_range < high_range:
                break
            else:
                print("Low value of range need to be smaller than high value of range")
        except ValueError:
            print("Please enter only numbers")

    return [low_range, high_range]


def filter_outliers(a_list: list[float], a_range: list[float]) -> list[float]:  # Filter numbers and return a new list
    new_list = []
    for number in a_list:
        if number >= a_range[0] and number <= a_range[1]:
            new_list.append(number)
    return new_list


def calc_stats(a_list: list[float]) -> tuple[float]:  # Calculate mean and sd
    mean = sum(a_list) / len(a_list)
    for i in range(len(a_list)):
        a_list[i] -= mean
        a_list[i] = a_list[i] ** 2
    sd = (1 / len(a_list) * sum(a_list)) ** (1 / 2)
    return mean, sd


def main():  # Main program
    filtered_list = filter_outliers(number_list_input(), range_input())
    print(f"Filtered list: {filtered_list}")
    stats = calc_stats(filtered_list)
    print(f"Mean: {stats[0]:.1f}")
    print(f"Standard Deviation: {stats[1]:.5f}")


if __name__ == "__main__":
    main()
```

```python
#################
#University of Toronto
#Faculty of Information
#Master of Information Program
#INF 1340H - Program for Data Science
#
#Student Name: Ruiqi
#Student Number: 1006743419
#Supervisor: Dr. Maher Elshakankiri
#
#
#Assignment 2, Problem 3
#Purpose: Word Cleaner
#Date Created: 10.1.2026
#Date Modified: 10.1.2026
#################
import sys  # To show error information


def read_words(file: str) -> list[str]:  # Read the file and store words in a list, seperate by space
    try:
        f = open(file, mode='r', encoding='utf-8')
    except OSError:
        print("Can not open ", file)
        if file == "words.txt":
            !wget https://raw.githubusercontent.com/feimiao3419/INF1340/refs/heads/main/Assignment%202/words.txt  # Download file from my github repository for sample run
    except:
        print("Unexpected error:", sys.exc_info()[0])
    else:
        word_list = []
        for line in f:
            word_list = word_list + line.lower().split()
        return word_list


def remove_punctuation(word: str) -> str:  # Remove select punctuation in text
    punctuation_list = [".", ",", ";", ":", "?", "!"]
    for punctuation in punctuation_list:
        word = word.replace(punctuation, "")
    return word


def check_other_character(word: str) -> bool:  # Check if the word is only contain alphabetic characters, apostrophes, or hyphens
    if word.replace("'", "").replace("-", "").isalpha():
        return True
    else:
        return False


def  remove_word_with_other_characters(word_list: list[str]) -> list[str]:  # Remove all words with other characters in list
    new_list = []
    for word in word_list:
        if check_other_character(word):
            new_list.append(word)
    return new_list


def main():  # Main function
    while True:
        file = input("Enter filename: ")
        word_list = read_words(file)
        if word_list != None:
            break
    for i in range(len(word_list)):
        word_list[i] = remove_punctuation(word_list[i])
    word_list = remove_word_with_other_characters(word_list)
    for word in word_list:
        print(word)


if __name__ == "__main__":
    main()
```

```python
#################
#University of Toronto
#Faculty of Information
#Master of Information Program
#INF 1340H - Program for Data Science
#
#Student Name: Ruiqi
#Student Number: 1006743419
#Supervisor: Dr. Maher Elshakankiri
#
#
#Assignment 2, Problem 4
#Purpose: Patient File and BMI
#Date Created: 10.2.2026
#Date Modified: 10.2.2026
#################
import pandas as pd
import sys
from typing import TextIO
from urllib.request import Request, urlopen, urlretrieve
from urllib.error import HTTPError, URLError
GITHUB_URL = "https://raw.githubusercontent.com/feimiao3419/INF1340/refs/heads/main/Assignment%202/"  # My github repository for this assignment


def check_github(file: str) -> None:  # Check if file in github and download
    for i in range(5):
        url = GITHUB_URL + file
        req = Request(url, method="HEAD")
        try:
            urlopen(req, timeout=10)
        except HTTPError:
            break
        except URLError:
            print("Error: Internet connect error, attempt: ", i+1, "of 5")
        else:
            urlretrieve(url, file)
            print(file, " downloaded from github, please try again.")
            break


def read_data(file: str) -> pd.DataFrame:  # Try read csv file to dataframe
    try:
        data = pd.read_csv(file)
    except OSError:
        print("Can not open ", file)
        check_github(file)
    except:
        print("Unexpected error:", sys.exc_info()[0])
    else:
        return data


def cal_result(patient: pd.DataFrame) -> tuple[str, float]:  # Calculate bmi if possible. -1 means height or weight not positive error, -2 means missing value error, -3 means other error
    try:
        height = float(patient[2])
        weight = float(patient[3])
        if height > 0 and weight > 0:
            bmi = weight * height ** -2 * 703
            return patient[1], bmi
        else:
            return patient[1], -1
    except ValueError:
        return patient[1], -2
    except:
        return patient[1], -3


def write_to_file(patient: tuple[str, float], file: TextIO) -> None:  # Output error message and write data to file
    if patient[1] > 0:
        print(f"{patient[0]} BMI: {patient[1]:.2f}")
        file.write(f"{patient[0]},{patient[1]:.2f}\n")
    elif patient[1] == -1:
        print(f"Error: Height and weight must be positive")
    elif patient[1] == -2:
        print(f"Error: Invalid height or weight for patient {patient[0]}")
    else:
        print(f"Error: Unknown error")


def main():
    while True:
        file = input("Enter filename: ")
        df = read_data(file)
        if df is not None:
            break
    new_file = open("bmi.txt", mode='w')
    for patient in df.itertuples():
        result = cal_result(patient)
        write_to_file(result, new_file)
    new_file.close()


if __name__ == "__main__":
    main()
```
