```python
from os import name
###############
#University of Toronto
#Faculty of Information
#Master of Information Program
#INF 1340H - Program for Data Science
#
#Student Name: Ruiqi
#Student Number: 1006743419
#Supervisor: Dr. Maher Elshakankiri
#
#
#Assignment 2, Problem 1
#Purpose: Purchases
#Date Create: 9.18.2026
#Date Modified: 9.22.2026
#################


def item_quantity_input() -> int:  # Request for item quantity input
    while True:
        quantity = input("Enter quantity: ")
        try:
            quantity = int(quantity)
            if quantity > 0:
                break
            else:
                print("Please input a positive integer.")
        except ValueError:
            print("Please input a positive integer.")
    return quantity


def unit_cost_input() -> float:  # Request for unit cost input
    while True:
        unit_cost = input("Enter unit cost: ")
        try:
            unit_cost = float(unit_cost)
            if unit_cost > 0:
                break
            else:
                print("Please input a positive number.")
        except ValueError:
            print("Please input a positive number.")
    return unit_cost


def is_taxable_input() -> bool:  # Request for is item taxable input
    while True:
        taxable = input("Is this item taxable (y/n)? ")
        if taxable in ("y", "n"):
            break
        else:
            print("Please input y or n.")
    if taxable == "y":
        return True
    else:
        return False


def each_item_input() -> list[int, float, bool]:  # Create a list for each item
    item_information = [item_quantity_input(), unit_cost_input(), is_taxable_input()]
    return item_information


def tax_rate_input() -> float:  # Request for tax rate
    while True:
        tax_rate = input("Enter sales tax rate (%): ")
        try:
            tax_rate = float(tax_rate)
            if tax_rate > 0:
                break
            else:
                print("Please input a positive number.")
        except ValueError:
            print("Please input a positive number.")
    return tax_rate


def number_of_items_input() -> int:  # Request for number of items
    while True:
        number_of_item = input("Enter number of different items: ")
        try:
            number_of_item = int(number_of_item)
            if number_of_item > 0:
                break
            else:
                print("Please input a positive integer.")
        except ValueError:
            print("Please input a positive integer.")
    return number_of_item


def main():  # Main program
    purchase_detail = []
    for i in range(1, number_of_items_input() + 1):
        item_name = "Item " + str(i)
        print(item_name)
        item_detail = [item_name] + each_item_input()
        purchase_detail.append(item_detail)
    tax_rate = tax_rate_input()
    total_cost = 0
    taxable_cost = 0
    for item_detail in purchase_detail:
        cost = item_detail[1] * item_detail[2]
        total_cost += cost
        if item_detail[3]:
            taxable_cost += cost
    tax = taxable_cost * 0.01 * tax_rate
    total_cost_after_tax = total_cost + tax
    print(f"Total cost (pre-tax): $ {total_cost:.2f}")
    print(f"Sales tax: $ {tax:.2f}")
    print(f"Total cost (with tax): $ {total_cost_after_tax:.2f}")


if __name__ == "__main__":
    main()
```
