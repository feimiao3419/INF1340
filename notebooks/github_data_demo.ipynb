# 示范：从 GitHub 下载数据并加载到 Colab

Demo: download a data file from GitHub and load it in Colab — no Google Drive mount needed.

本 notebook 演示三种方法，任选其一即可：
1. **pandas 直接读 raw URL**（最简单，推荐）
2. **wget 下载到 Colab 本地再读**
3. **git clone 克隆整个仓库**（文件多的时候用）

## 方法 1：pandas 直接读 raw URL（推荐）

GitHub 上每个文件都有一个 raw 地址（在文件页面点 **Raw** 按钮可以复制）。
`pd.read_csv()` 可以直接读 URL，连下载步骤都省了。

```python
import pandas as pd

url = "https://raw.githubusercontent.com/feimiao3419/INF1340/main/data/example.csv"
df = pd.read_csv(url)
df
```

## 方法 2：wget 下载到 Colab 本地再读

适合需要先拿到文件、或之后还要对文件做其他操作的场景。
注意：Colab 会话断开后，下载的文件会消失，重新运行单元格即可。

```python
!wget -q https://raw.githubusercontent.com/feimiao3419/INF1340/main/data/example.csv

df2 = pd.read_csv("example.csv")  # 读本地文件
df2.head()
```

## 方法 3：git clone 克隆整个仓库

数据文件多、或者还需要仓库里的代码时，一次克隆全部。

```python
!git clone -q https://github.com/feimiao3419/INF1340.git

df3 = pd.read_csv("INF1340/data/example.csv")
df3.head()
```

## 换成你自己的数据

1. 把文件上传到仓库的 `data/` 文件夹（GitHub 网页上 Add file → Upload files 拖拽即可）
2. 把方法 1 里的 URL 最后的 `example.csv` 改成你的文件名
3. 运行！

提示：仓库是 public 的，不要上传含个人隐私的数据。
